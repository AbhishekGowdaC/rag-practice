```python
sample_texts={
    "../data/text_files/AI_intro.txt":"""Introduction to Artificial Intelligence

Artificial Intelligence (AI) is a branch of computer science that focuses on creating machines and software capable of performing tasks that typically require human intelligence. These tasks include learning, problem-solving, decision-making, language understanding, and image recognition.

Machine Learning

Machine Learning (ML) is a subset of Artificial Intelligence that enables computers to learn patterns from data and make predictions without being explicitly programmed for every task.

There are three main types of machine learning: supervised learning, unsupervised learning, and reinforcement learning.

Supervised learning uses labeled data to train models. Examples include spam detection and house price prediction.

Unsupervised learning identifies patterns in unlabeled data. Examples include customer segmentation and clustering.

Reinforcement learning trains models through interactions with an environment using rewards and penalties.

Deep Learning

Deep Learning is a subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns. It is widely used in computer vision, speech recognition, natural language processing, and generative AI.

Natural Language Processing

Natural Language Processing (NLP) enables computers to understand, interpret, and generate human language. Common NLP applications include chatbots, sentiment analysis, language translation, text summarization, and question answering.

Retrieval-Augmented Generation

Retrieval-Augmented Generation (RAG) is an AI architecture that combines information retrieval with a large language model (LLM). Instead of relying only on knowledge learned during training, the model retrieves relevant information from an external knowledge source before generating an answer.

A typical RAG pipeline consists of document loading, text splitting, embedding generation, vector storage, retrieval, and response generation.

Document loading reads documents from text files, PDFs, websites, or other sources.

Text splitting divides large documents into smaller chunks so that relevant information can be retrieved efficiently.

Embedding generation converts text chunks into numerical vector representations that capture semantic meaning.

Vector storage saves embeddings in a vector database for efficient searching.

Retrieval identifies the chunks most relevant to a user's question.

Generation provides the retrieved context to a language model, which uses it to generate an answer.

Embeddings

Embeddings are numerical representations of text, images, or other data. Text embeddings capture semantic relationships and help systems identify passages with similar meanings even when they use different words.

For example, the sentences "How can I learn Python?" and "What is the best way to study Python programming?" may have similar embeddings because they express related meanings.

Vector Databases

A vector database stores and searches vector embeddings. It helps AI applications retrieve information based on semantic similarity rather than relying only on exact keyword matches.

Popular vector databases and vector search tools include FAISS, Chroma, Pinecone, and Qdrant.

Benefits of RAG

RAG can provide answers based on external documents. It can reduce unsupported answers when retrieval and generation are implemented correctly. Documents can be updated without retraining the entire language model. Retrieved passages can help users verify the source of an answer. RAG can also support question answering over private or specialized knowledge bases.

Example Use Case

Imagine a university has hundreds of documents containing course details, examination rules, and attendance policies.

A student asks, "What is the minimum attendance required for an examination?"

A RAG application searches the university documents, retrieves the relevant attendance policy, and supplies that information to the language model. The model then generates an answer based on the retrieved passage.

This approach is useful because the answer can be grounded in the university's actual documents instead of relying entirely on the model's general knowledge.

Conclusion

Artificial Intelligence, Machine Learning, Deep Learning, NLP, and RAG are important technologies in modern AI applications. RAG is particularly useful for building document-based question-answering systems that combine semantic search with language generation."""
}
```

```python
import os
for filepath,content in sample_texts.items():
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("Sample text file created ")
```

```python
from langchain_community.document_loaders import TextLoader


loader=TextLoader("../data/text_files/AI_intro.txt", encoding="utf-8")

document = loader.load()
print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", #pattern to match files
    loader_cls=TextLoader,
    loader_kwargs={'encoding':'utf-8'},

)

document = dir_loader.load()
document
```
